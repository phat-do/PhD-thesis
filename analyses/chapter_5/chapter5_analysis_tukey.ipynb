```python
import pandas as pd
import statsmodels.api as sm
import statsmodels.stats.multicomp as mc

df = pd.read_csv("chapter5_data.csv")
cer = df[df['metric'] == 'cer']
mos = df[df['metric'] == 'mos']
```

```python
interaction_groups = "Input_" + cer.input.astype(str) + " & " + "Dict_" + cer.dict.astype(str)

comp = mc.MultiComparison(cer["value"], interaction_groups)
post_hoc_res = comp.tukeyhsd()
post_hoc_res.summary()
```

```python
interaction_groups = "Input_" + mos.input.astype(str) + " & " + "Dict_" + mos.dict.astype(str)

comp = mc.MultiComparison(mos["value"], interaction_groups)
post_hoc_res = comp.tukeyhsd()
post_hoc_res.summary()
```
