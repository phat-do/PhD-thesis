```python
import pandas as pd
from scipy import stats
```

# Load data

```python
metadata = pd.read_csv('chapter4_data.csv')
zero = metadata[metadata['amount']==0] # subset the zeroshot data
```

# Q4a (Section 4.3.1)

The following cells correspond to Table 4.1

```python
# p-value
print("p-value of Wilcoxon ranked test that MSE: BVCC-SOMOS << SOMOS (Utterance)")
print(stats.wilcoxon(zero[(zero['model']=='BVCC-SOMOS') & (zero['level']=='Utterance') & (zero['metric']=='MSE')]["value"], 
                    zero[(zero['model']=='SOMOS') & (zero['level']=='Utterance') & (zero['metric']=='MSE')]["value"], 
                    alternative='less').pvalue)
print("p-value of Wilcoxon ranked test that MSE: BVCC-SOMOS >> SOMOS (System)")
print(stats.wilcoxon(zero[(zero['model']=='BVCC-SOMOS') & (zero['level']=='System') & (zero['metric']=='MSE')]["value"], 
                    zero[(zero['model']=='SOMOS') & (zero['level']=='System') & (zero['metric']=='MSE')]["value"], 
                    alternative='less').pvalue)
print("p-value of Wilcoxon ranked test that MSE: BVCC-SOMOS << BVCC (System)")
print(stats.wilcoxon(zero[(zero['model']=='BVCC-SOMOS') & (zero['level']=='System') & (zero['metric']=='MSE')]["value"], 
                    zero[(zero['model']=='BVCC') & (zero['level']=='System') & (zero['metric']=='MSE')]["value"], 
                    alternative='less').pvalue)
```

```python
# p-value
print("p-value of Wilcoxon ranked test that LCC: BVCC-SOMOS >> SOMOS (Utterance)")
subset = zero[(zero['model'].isin(['BVCC-SOMOS','SOMOS'])) & (zero['level']=='Utterance') & (zero['metric']=='LCC')]
print(stats.wilcoxon(subset[(subset['model']=='BVCC-SOMOS')]["value"],
                    subset[(subset['model']=='SOMOS')]["value"],
                    alternative='greater').pvalue)
print("p-value of Wilcoxon ranked test that LCC: BVCC-SOMOS >> SOMOS (System)")
subset = zero[(zero['model'].isin(['BVCC-SOMOS','SOMOS'])) & (zero['level']=='System') & (zero['metric']=='LCC')]
print(stats.wilcoxon(subset[(subset['model']=='BVCC-SOMOS')]["value"],
                    subset[(subset['model']=='SOMOS')]["value"],
                    alternative='greater').pvalue)
```

```python
# p-value
print("p-value of Wilcoxon ranked test that SRCC: BVCC-SOMOS >> SOMOS (Utterance)")
subset = zero[(zero['model'].isin(['BVCC-SOMOS','SOMOS'])) & (zero['level']=='Utterance') & (zero['metric']=='SRCC')]
print(stats.wilcoxon(subset[(subset['model']=='BVCC-SOMOS')]["value"],
                    subset[(subset['model']=='SOMOS')]["value"],
                    alternative='greater').pvalue)
print("p-value of Wilcoxon ranked test that SRCC: BVCC-SOMOS >> BVCC (System)")
subset = zero[(zero['model'].isin(['BVCC-SOMOS','BVCC'])) & (zero['level']=='System') & (zero['metric']=='SRCC')]
print(stats.wilcoxon(subset[(subset['model']=='BVCC-SOMOS')]["value"],
                    subset[(subset['model']=='BVCC')]["value"],
                    alternative='greater').pvalue)
```

# Q4b (Section 4.3.2)

```python
# only BVCC-SOMOS

alt = {'MSE': 'greater', 'LCC': 'less', 'SRCC': 'less'}

for metric in ['MSE', 'LCC', 'SRCC']:
    for level in ['Utterance', 'System']:
        subset = metadata[(metadata['model']=='BVCC-SOMOS') & (metadata['level']==level) & (metadata['metric']==metric)]
        print(f"Current: level {level}, metric {metric}:")
        for amount in [10, 20, 30, 40]:
            pvalue = stats.wilcoxon(subset[(subset['amount']==amount-10)]["value"],
            subset[(subset['amount']==amount)]["value"],
            alternative=alt[metric]).pvalue
            print(f"\tSignificance: {amount-10} {alt[metric]} than {amount}: {True if pvalue <= 0.05 else False}, {pvalue}")
```
